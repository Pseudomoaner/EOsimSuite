```python
#Import packages used
import matplotlib
import matplotlib.pyplot as plt
import numpy as np
from scipy.integrate import solve_ivp
from importlib import reload
from cycler import cycler

#Reload custom modules if being tested
import sys
sys.path.append('C:\\Users\\bi1ojm\\Documents\\GitHub\\FeedOrFold\\EOsimSuite\\')
from EOsimsuite import EOsimsuite as EOsim
```

```python
#Sensitivity function - rs are resource concs, specInd is the index of the species under consideration, C is the consumption matrix, delW is the resource values and D is the mortality rate
def sensFunc(rs,specInd,EOparams):
    ws = EOparams['ws']
    ps = EOparams['ps']
    
    return sum([rs[rho]*ps[specInd,rho]*ws[rho] for rho in range(len(rs))])

def gradSensFunc(rs,specInd,EOparams):
    ws = EOparams['ws']
    ps = EOparams['ps']

    return(np.array([ps[specInd,rho]*ws[rho] for rho in range(len(rs))]))

#Impact function for a single species
def impactFunc(rs,specInd,EOparams):
    ps = EOparams['ps']
    
    impact = np.array([-ps[specInd,rho]*rs[rho] for rho in range(len(rs))])

    return impact

def instInteract(rs,alphInd,betInd,EOparams):
    return(np.dot(impactFunc(rs,betInd,EOparams),gradSensFunc(rs,alphInd,EOparams)))

def intrinsicGR(rs,specInd,EOparams,sysParams):
    vs = sysParams['vs']
    Ks = sysParams['Ks']
    ts = sysParams['ts']

    ps = EOparams['ps']
    ws = EOparams['ws']
    
    intGR = sum([ws[rho]*ps[specInd,rho]*rs[rho]*vs[rho]*(Ks[rho] - rs[rho])/Ks[rho] for rho in range(len(rs))])
    
    return intGR
```

```python
#Functions that generate random C and D matrices
def genRandMats(noSpec,noRes):
    Pvar = 0.5 #Variation of resource preferences
    Pmin = 0 #Minimum resource preference
    Wvar = 0.5
    Wmin = 0
    
    ps = Pmin + np.random.rand(noSpec,noRes)*Pvar
    ws = Wmin + np.random.rand(noRes)*Wvar
    
    return (ps,ws)
```

```python
noSpec = 2
noRes = 2

Ks = np.array([2,3])#,4,5])
vs = np.array([0.8,1.2])#,1,1.1])
ts = np.array([0.1,0.08])#,0.12,0.13])

sysParams = {'Ks':Ks,'vs':vs,'ts':ts}

B0 = np.ones((noSpec,))*0.5 #Starting concentration of bacteria in batch (biomass)
R0 = np.ones((noRes,))

(ps,ws) = genRandMats(noSpec,noRes)
EOparams = {'ps':ps,'ws':ws}

coexists = 0

while (coexists < noSpec):
    (ps,ws) = genRandMats(noSpec,noRes)
    EOparams = {'ps':ps,'ws':ws}
    
    macSimBase = EOsim(impactFunc,sensFunc,EOparams,sysParams,B0,R0)
    macSimBase.set_timeSettings(tSpan = [0,1000], dt=0.1)
    macSimBase.set_instInteract(instInteract)
    macSimBase.set_intrinsicGR(intrinsicGR)

    macSimBase.simEOmodel('MacArthur')

    coexists = sum(macSimBase.specTimeseries[:,-1] > 0.1)

macSimBase.plotSolutionTimecourse()
```

```python
#Sweep over different starting resources
for R1 in range(5):
    for R2 in range(5):
        macSim = EOsim(impactFunc,sensFunc,EOparams,sysParams,B0,np.array([R1/5+0.1,R2/5+0.1]))
        macSim.set_timeSettings(tSpan = [0,1000], dt=0.1)
        macSim.set_instInteract(instInteract)

        macSim.simEOmodel('MacArthur')

        plt.plot(macSim.resTimeseries[0,:],macSim.resTimeseries[1,:],'r')
```

```python
#gLV mapping with timescale separation
def getAts(ps,ws,Ks,vs,alphInd,betInd):
    aTS = -sum([ps[alphInd,rhoInd]*ps[betInd,rhoInd]*ws[rhoInd]*Ks[rhoInd]/vs[rhoInd] for rhoInd in range(len(ws))])
    return aTS
    
def getMuTs(ps,ws,Ks,ts,alphInd):
    muTS = sum([Ks[rhoInd]*ws[rhoInd]*ps[alphInd,rhoInd] for rhoInd in range(len(ws))]) - ts[alphInd]
    return muTS

Ats = np.array([[getAts(ps,ws,Ks,vs,alphInd,betInd) for alphInd in range(len(ws))] for betInd in range(len(ws))])
muTs = np.array([getMuTs(ps,ws,Ks,ts,alphInd) for alphInd in range(len(ws))])

print(Ats)
print(muTs)
```

```python
#gLV mapping with EO method
def getAeos(rs,noSpec,EOparams):
    return(np.array([[instInteract(rs,alphInd,betInd,EOparams)
                      for betInd in range(noSpec)]
                     for alphInd in range(noSpec)]))

def getGAs(rs,noSpec,EOparams,sysParams):
    return(np.array([intrinsicGR(rs,specInd,EOparams,sysParams)
                     for specInd in range(noSpec)]))

invMat = np.array([[ws[rhoInd] * ps[alphInd,rhoInd] for alphInd in range(len(ws))] for rhoInd in range(len(ws))])
rStar = np.matmul(ts,np.linalg.inv(invMat))

macSimBase.set_intrinsicGR(intrinsicGR)

AEO = getAeos(rStar,noSpec,EOparams)
muEO = getGAs(rStar,noSpec,EOparams,sysParams)

print(AEO)
print(muEO)
```

```python
def gLV(t, y, intMat, GRs):
    pcgrs = GRs + np.sum(np.matmul(np.diag(y),intMat),axis=0)
    return np.multiply(pcgrs,y)

def simGLV(intMat,GRs,x0):
    tMax = 5000
    dt = 1
    tSteps = np.arange(0,tMax,dt)
    solution = solve_ivp(fun=gLV,t_span=[0,tMax],y0=x0,args=(intMat,GRs),t_eval=tSteps)
    return solution.y

dX0 = xStar + np.array([1.5,0.3])#,0.1,0.2]);

GLVeo = simGLV(AEO,muEO,dX0)
GLVts = simGLV(Ats,muTs,dX0)

macSimPeterb = EOsim(impactFunc,sensFunc,EOparams,sysParams,dX0,R0)
macSimPeterb.set_timeSettings(tSpan = [0,1000], dt=0.1)
macSimPeterb.set_instInteract(instInteract)

macSimPeterb.simEOmodel('MacArthur')

fig = matplotlib.pyplot.gcf()
fig.set_size_inches(2, 2)

plt.plot(GLVeo[0,:],GLVeo[1,:],'b')
plt.plot(GLVts[0,:],GLVts[1,:],'k')
plt.plot(macSimPeterb.specTimeseries[0,:],macSimPeterb.specTimeseries[1,:],'r')

'''
dX0 = xStar + np.array([-0.5,-0.2,0.1,0.2]);

GLVeo = simGLV(AEO,muEO,dX0)
GLVts = simGLV(Ats,muTs,dX0)

macSimPeterb = EOsim(impactFunc,sensFunc,EOparams,sysParams,dX0,R0)
macSimPeterb.set_timeSettings(tSpan = [0,1000], dt=0.1)
macSimPeterb.set_instInteract(instInteract)

macSimPeterb.simEOmodel('MacArthur')

plt.plot(GLVeo[0,:],GLVeo[1,:],'b')
plt.plot(GLVts[0,:],GLVts[1,:],'k')
plt.plot(macSimPeterb.specTimeseries[0,:],macSimPeterb.specTimeseries[1,:],'r')

plt.plot(macSimPeterb.specTimeseries[0,-1],macSimPeterb.specTimeseries[1,-1],'ko')

#plt.axis('equal')
#plt.axis([0,3.5,0,3.5])

#plt.savefig("C:\\Users\\bi1ojm\\Desktop\\FeedyFoldy\\GlvTS.pdf")
'''
```

```python
#Separation of timescales approximation of resource abundances
def ssRs(EOparams,sysParams,xs):
    Ks = sysParams['Ks']
    vs = sysParams['vs']
    ps = EOparams['ps']
    
    rsOut = [Ks[rho] - (Ks[rho]/vs[rho])*sum([ps[bet,rho]*xs[bet] for bet in range(len(xs))]) for rho in range(len(Ks))]
    return(np.array(rsOut))

def gLVaccel(t, y, EOparams, sysParams):
    noSpec = len(sysParams['Ks'])
    x = y[:noSpec]
    dx = y[noSpec:]

    #Get current GLV-like params based on ss-approximation of current system species composition
    currRs = ssRs(EOparams,sysParams,x)
    Aeo = getAeos(rStar,noSpec,EOparams)
    muEO = getGAs(rStar,noSpec,EOparams,sysParams)
    
    GLVbit = muEO + np.sum(np.matmul(np.diag(x),Aeo),axis=0)
    
    ddx = np.multiply(GLVbit,x) + np.divide(np.multiply(dx,dx),x)# - dx/20
    return np.concatenate((dx,ddx), axis=0)

def simGLVaccel(EOparams,sysParams,x0,dx0):
    tMax = 200
    dt = 0.1
    tSteps = np.arange(0,tMax,dt)    
    
    solution = solve_ivp(fun=gLVaccel,t_span=[0,tMax],y0=np.concatenate((x0,dx0),axis=0),args=(EOparams,sysParams),t_eval=tSteps)
    return solution.y

dX0 = xStar + np.array([-0.5,0.3])#,0.1,0.2]);
ddX0 = muEO + np.sum(np.matmul(np.diag(dX0),AEO))

GLVeoAccel = simGLVaccel(EOparams,sysParams,dX0,ddX0)
GLVeo = simGLV(AEO,muEO,dX0)

macSimPeterb = EOsim(impactFunc,sensFunc,EOparams,sysParams,dX0,R0)
macSimPeterb.set_timeSettings(tSpan = [0,1000], dt=0.1)
macSimPeterb.set_instInteract(instInteract)

macSimPeterb.simEOmodel('MacArthur')

fig = matplotlib.pyplot.gcf()
fig.set_size_inches(2, 2)

#plt.plot(GLVeo[0,:],GLVeo[1,:],'b')
plt.plot(macSimPeterb.specTimeseries[0,:],macSimPeterb.specTimeseries[1,:],'r')
plt.plot(GLVeoAccel[0,:],GLVeoAccel[1,:],'g')
'''
dX0 = xStar + np.array([-0.5,-0.2,0.1,0.2]);

GLVeoAccel = simGLVaccel(EOparams,sysParams,dX0)
GLVeo = simGLV(AEO,muEO,dX0)

macSimPeterb = EOsim(impactFunc,sensFunc,EOparams,sysParams,dX0,R0)
macSimPeterb.set_timeSettings(tSpan = [0,1000], dt=0.1)
macSimPeterb.set_instInteract(instInteract)

macSimPeterb.simEOmodel('MacArthur')

plt.plot(macSimPeterb.specTimeseries[0,:],macSimPeterb.specTimeseries[1,:],'r')
plt.plot(GLVeoAccel[0,:],GLVeoAccel[1,:],'g')

plt.plot(macSimPeterb.specTimeseries[0,-1],macSimPeterb.specTimeseries[1,-1],'ko')

#plt.axis('equal')
#plt.axis([0,3.5,0,3.5])

plt.savefig("C:\\Users\\bi1ojm\\Desktop\\FeedyFoldy\\Damped.pdf")
'''
```
