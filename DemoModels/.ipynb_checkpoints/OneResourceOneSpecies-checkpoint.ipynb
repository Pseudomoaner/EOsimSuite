```python
#Import packages used
import matplotlib
import matplotlib.pyplot as plt
import numpy as np
from scipy.integrate import solve_ivp
```

```python
def sensFunc(rs,specInd,EOparams):
    a = EOparams['a']

    return a*rs

def impactFunc(rs,specInd,EOparams):
    a = EOparams['a']

    return -a*rs

def chemostat(t, y, EOparams, sysParams, impactFunc, sensFunc):
    rs = y[0]
    specs = y[1]

    dr = impactFunc(rs,0,EOparams)*specs
    dr += (sysParams['rIn'] - rs)*sysParams['D'] #Sigma for a chemostat   
    ds = (sensFunc(rs,0,EOparams) - sysParams['D'])*specs

    dY = np.array([dr,ds])
    dY.shape = (2,)
    return dY
```

```python
EOparams = {'a':0.8}

sysParams = {'rIn':1,'D':0.1}

maxChemT = 20
dilRate = 0.1

inArgs = (EOparams,sysParams,impactFunc,sensFunc)

tSteps = np.arange(0,maxChemT,0.01)

#SS solution has been peturbed by addition of 0.1 of species abundance.
xStar = 0.875
rStar = 0.125
solution = solve_ivp(fun=chemostat, t_span=[0,maxChemT], y0=np.array([rStar,xStar+0.1]), t_eval=tSteps, args=inArgs)
```

```python
plt.plot(solution.y[1,:]) #Abundance over time
```

```python
plt.plot(sensFunc(solution.y[0,:],0,EOparams)) #PCGR
```

```python
plt.plot(np.divide(np.diff(sensFunc(solution.y[0,:],0,EOparams))*10,solution.y[0,0:-1])) #PCGA
```

```python
def gLVaccel(t, y, intMat, GAs):
    x = y[:len(GAs)]
    dx = y[len(GAs):]
    pcgas = GAs + np.sum(np.matmul(np.diag(x),intMat),axis=0)
    ddx = np.multiply(pcgas,x)
    return np.concatenate((dx,ddx), axis=0)

def simGLVaccel(intMat,GAs,x0):
    tMax = 500
    dt = 0.1
    tSteps = np.arange(0,tMax,dt)
    solution = solve_ivp(fun=gLVaccel,t_span=[0,tMax],y0=np.concatenate((x0,np.zeros(len(x0),)),axis=0),args=(intMat,GAs),t_eval=tSteps)
    return solution.y

def gLVaccelR(t, y, intMat, GAs):
    x = y[:len(GAs)]
    dx = y[len(GAs):]
    pcgas = GAs + np.sum(np.matmul(np.diag(x),intMat),axis=0)
    ddx = np.multiply(pcgas,(x**2)/(x-1))
    return np.concatenate((dx,ddx), axis=0)

def simGLVaccelR(intMat,GAs,x0):
    tMax = 500
    dt = 0.1
    tSteps = np.arange(0,tMax,dt)
    solution = solve_ivp(fun=gLVaccel,t_span=[0,tMax],y0=np.concatenate((x0,np.zeros(len(x0),)),axis=0),args=(intMat,GAs),t_eval=tSteps)
    return solution.y

GLVeoAccel = simGLVaccel(-a**2,,dX0)

plt.plot(GLVeo[0,:],GLVeo[1,:],'b')
plt.plot(GLVeoAccel[0,:],GLVeoAccel[1,:],'k')
```
