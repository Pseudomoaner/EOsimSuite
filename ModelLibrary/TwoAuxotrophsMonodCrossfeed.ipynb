```python
#Sensitivity function - rs are resource concs, specInd is the index of the species under consideration, Kms are the half-maximal velocity concentrations and fs are the investment fractions/leakage penalties
def sensFunc(rs,specInd,EOparams):
    Kms = EOparams['Kms']
    Vmax = EOparams['Vmax']
    fs = EOparams['fs']

    monodG = rs[2]/(rs[2]+Kms[2])
    
    if specInd == 0:
        monodP = rs[0]/(rs[0] + Kms[0])
        sens = Vmax*(1-fs[specInd])*monodP*monodG
    elif specInd == 1:
        monodW = rs[1]/(rs[1] + Kms[0])
        sens = Vmax*(1-fs[specInd])*monodW*monodG
    
    return sens

#Gradient of the sensitivity function in the resource space
def gradSensFunc(rs,specInd,EOparams):
    Kms = EOparams['Kms']
    Vmax = EOparams['Vmax']
    Ys = EOparams['Ys']
    fs = EOparams['fs']
    
    sensGrad = np.zeros(rs.shape)
    
    monodG = rs[2]/(rs[2]+Kms[2])
    dMonodG = Kms[2]/(rs[2] + Kms[2])**2
    
    if specInd == 0:
        monodP = rs[0]/(rs[0] + Kms[0])
        dMonodP = Kms[0]/(rs[0] + Kms[0])**2
        sensGrad[0] = dMonodP*monodG*Vmax*(1-fs[specInd])
        sensGrad[2] = dMonodG*monodP*Vmax*(1-fs[specInd])
    elif specInd == 1:
        monodW = rs[1]/(rs[1] + Kms[1])
        dMonodW = Kms[1]/(rs[1] + Kms[1])**2
        sensGrad[1] = dMonodW*monodG*Vmax*(1-fs[specInd])
        sensGrad[2] = dMonodG*monodW*Vmax*(1-fs[specInd])
    
    return sensGrad

#Impact function for a single species
def impactFunc(rs,specInd,EOparams):
    Kms = EOparams['Kms']
    Vmax = EOparams['Vmax']
    Ys = EOparams['Ys']
    fs = EOparams['fs']
    
    impact = np.zeros(rs.shape)
    
    monodG = rs[2]/(rs[2]+Kms[2])
    
    if specInd == 0: #The proline consumer/tryptophan producer
        monodP = rs[0]/(rs[0] + Kms[0])
        impact[0] = -(Vmax/Ys[0])*monodG*monodP
        impact[1] = fs[specInd]*Vmax*monodG*monodP
        impact[2] = -(Vmax/Ys[2])*monodG*monodP
    elif specInd == 1: #The tryptophan consumer/proline producer
        monodW = rs[1]/(rs[1] + Kms[1])
        impact[0] = fs[specInd]*Vmax*monodG*monodW
        impact[1] = -(Vmax/Ys[1])*monodG*monodW
        impact[2] = -(Vmax/Ys[2])*monodG*monodW
    
    return impact
```
